```python
plt.figure(figsize=(25,6))
alpha=0.6
aF1.avg_forman.plot(marker=" ",alpha=alpha,lw=8,label="d=1 (edges)")
plt.axvline(x = 1/2, color = 'lightgray', ls="dashed")
plt.axvline(x = np.sqrt(5)/4, color = 'lightgray', ls="dashed")
plt.axvline(x = np.sqrt(2)/2, color = 'lightgray', ls="dashed")
plt.axvline(x = 1.0, color = 'lightgray', ls="dashed")
plt.axvline(x = 5/4, color = 'lightgray', ls="dashed")
plt.axvline(x = np.sqrt(17)/4, color = 'lightgray', ls="dashed")
plt.axvline(x = np.sqrt(37)/4, color = 'lightgray', ls="dashed")
#plt.axvline(x = np.sqrt(37)/4, color = 'lightgray', ls="dashed")
plt.axvline(x =2.0, color = 'lightgray', ls="dashed")
 
aF2.avg_forman.plot(marker=" ",lw=7,alpha=alpha,label="d=2 (triangles)")
aF3.avg_forman.plot(marker=" ",lw=5.,alpha=alpha,label="d=3 (tetrahedros)")
plt.legend(framealpha=0,fontsize=15)
plt.xticks(xticks,xlabels,fontsize=20)
plt.yticks(fontsize=20)
plt.xlabel("Cutoff distance (r)",fontsize=20)
plt.ylabel("Average Forman-Ricci Curvature",fontsize=20)
sns.despine(top=True,right=True)
plt.savefig("FRC_filtration_network.png",dpi=150,bbox_inches="tight")
```

```python
def analytic(n,p,d):
    return (d+2)*(n-(d+1))*p**(d+1) +(d+1)-(d+1)*(n-(d+1))*p**d
```

```python
aF1.cutoff
```

```python
plt.plot(aF1.cutoff,[analytic(100,p,1) for p in aF1.cutoff])
```

```python
plt.figure(figsize=(12,9))
aF1.avg_forman.plot(marker=".")
plt.plot(aF1.cutoff,[analytic(100,p,1) for p in aF1.cutoff],color="gray")
aF2.avg_forman.plot(marker=".")
plt.plot(aF2.cutoff,[analytic(100,p,2) for p in aF2.cutoff],color="gray")
```
